```python
query = """
SELECT LaunchSite, COUNT(*) AS Number_of_Launches
FROM SPACEXTBL
GROUP BY LaunchSite
ORDER BY Number_of_Launches DESC;
"""

launch_site_analysis = pd.read_sql_query(query, connection)

launch_site_analysis
```

```python
query = """
SELECT LaunchSite,
       SUM(PayloadMass) AS Total_Payload_Mass
FROM SPACEXTBL
GROUP BY LaunchSite
ORDER BY Total_Payload_Mass DESC;
"""

payload_analysis = pd.read_sql_query(query, connection)

payload_analysis
```

```python
query = """
SELECT LaunchSite,
       COUNT(*) AS Total_Launches,
       SUM(Class) AS Successful_Launches,
       ROUND(100.0 * SUM(Class) / COUNT(*), 2) AS Success_Rate
FROM SPACEXTBL
GROUP BY LaunchSite
ORDER BY Success_Rate DESC;
"""

success_rate_analysis = pd.read_sql_query(query, connection)

success_rate_analysis
```
